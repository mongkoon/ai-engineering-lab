# Local environment and data inspection

This clean notebook demonstrates environment checks without printing cloud account identifiers or credentials. Source code and recorded outputs are in `src/` and `results/`. No AWS jobs are started.

```python
import sys
import sklearn
import pandas as pd
from pathlib import Path

root = Path.cwd()
if not (root / "src").exists() and (root.parent / "src").exists():
    root = root.parent
print("Python:", sys.version.split()[0])
print("scikit-learn:", sklearn.__version__)
print("Project root:", root.name)
print("Preprocessing script exists:", (root / "src/preprocess.py").is_file())
print("Evaluation script exists:", (root / "src/evaluate.py").is_file())
```

```python
from sklearn.datasets import load_breast_cancer
frame = load_breast_cancer(as_frame=True).frame
print("Dataset shape:", frame.shape)
print("Target distribution:", frame.target.value_counts().sort_index().to_dict())
print("Target 0 = malignant; target 1 = benign")
```
